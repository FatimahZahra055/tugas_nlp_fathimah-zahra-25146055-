## 1. Muat Dataset & Informasi

```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset = load_dataset("dipawidia/ecommerce-product-reviews-sentiment", split="train")
df = dataset.to_pandas()
print("Jumlah data:", len(df))
print("\nKolom:", list(df.columns))
print("\nTipe data:")
print(df.dtypes)
print("\n5 baris pertama:")
print(df.head())
```

## 2. Distribusi Label Sentimen

```python
sentimen_counts = df['sentimen'].value_counts().sort_index()
print("Distribusi label sentimen:")
print(sentimen_counts)
print("\nPersentase:")
print((sentimen_counts / len(df) * 100).round(2))

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(sentimen_counts.index.astype(str), sentimen_counts.values, color=['steelblue', 'orange'])
ax.set_xlabel('Sentimen (0 = Negatif, 1 = Positif)')
ax.set_ylabel('Jumlah')
ax.set_title('Distribusi Label Sentimen')
for i, v in enumerate(sentimen_counts.values):
    ax.text(i, v + 50, str(v), ha='center')
plt.tight_layout()
plt.show()
```

## 3. Preprocessing Teks

```python
import re
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords', quiet=True)

def case_folding(text):
    return str(text).lower()

def cleaning(text):
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    text = re.sub(r'@\w+|#\w+', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def tokenisasi(text):
    return re.findall(r'[a-z]+', text)

stop_words = set(stopwords.words('indonesian'))

def stopword_removal(tokens):
    return [t for t in tokens if t not in stop_words]

from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
stemmer = StemmerFactory().create_stemmer()

def stemming(tokens):
    return [stemmer.stem(t) for t in tokens]

def preprocess(text):
    t = case_folding(text)
    t = cleaning(t)
    toks = tokenisasi(t)
    toks = stopword_removal(toks)
    toks = stemming(toks)
    return ' '.join(toks), toks

sample_df = df.head(10).copy()
results = sample_df['review'].apply(preprocess)
sample_df['review_clean'], sample_df['tokens'] = zip(*results)

print("=== 10 Contoh Sebelum & Sesudah Preprocessing ===")
for i, row in sample_df.iterrows():
    print("---", i, "Label:", row['sentimen'])
    print("Sebelum:", row['review'])
    print("Sesudah:", row['review_clean'])
    print("Token:", row['tokens'])
```

## 4. Pengaruh Preprocessing terhadap Kualitas Data

```python
import statistics

df['review_len_before'] = df['review'].apply(lambda x: len(str(x).split()))
df['review_clean'] = df['review'].apply(lambda x: preprocess(x)[0])
df['review_len_after'] = df['review_clean'].apply(lambda x: len(x.split()))

print("Statistik panjang teks (jumlah kata):")
print("Sebelum preprocessing:")
print(df['review_len_before'].describe())
print("\nSesudah preprocessing:")
print(df['review_len_after'].describe())
```

## 5. Kesimpulan
Preprocessing membantu membersihkan teks dari noise (URL, angka, tanda baca, stopword) dan mengubah kata ke bentuk dasar (stemming), sehingga data teks lebih konsisten dan siap untuk pelatihan model klasifikasi sentimen.
